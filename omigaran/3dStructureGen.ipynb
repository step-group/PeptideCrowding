```python
import nglview
```

```python
# Show protein
omigaran = "Structure2D_COMPOUND_CID_16131445.sdf"
view = nglview.show_structure_file(omigaran)
view.add_representation(repr_type='ball', selection='all')
view._remote_call('setSize', target='Widget', args=['','600px'])
view
```

```python
from openbabel import openbabel as ob
```

```python
mol = ob.OBMol()
conv = ob.OBConversion()
format = conv.FormatFromExt(omigaran)
conv.SetInAndOutFormats(format, format)
conv.ReadFile(mol, omigaran)

ff = ob.OBForceField.FindForceField("MMFF94")
ff.Setup(mol)

ff.SteepestDescent(500)
ff.Energy()
```

```python
ff.SystematicRotorSearch(20)
ff.Energy()
```
